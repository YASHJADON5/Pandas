```python
data = {
    "Department": ["IT", "IT", "HR", "HR", "Sales", "Sales", "IT", "Sales"],
    "Employee": ["A", "B", "C", "D", "E", "F", "G", "H"],
    "Salary": [50000, 60000, 45000, 55000, 40000, 50000, 70000, 45000],
    "Experience": [2, 4, 1, 3, 2, 5, 6, 3]
}

df = pd.DataFrame(data)

grouped = df.groupby("Department").agg(
    Total_Salary=pd.NamedAgg(column="Salary", aggfunc="mean"))


grouped_1 = df.groupby("Department").agg(
    Total_Salary=pd.NamedAgg(column="Salary", aggfunc="sum"),
    Average_Experience=pd.NamedAgg(column="Experience", aggfunc="mean")
)

grouped_2 = df.groupby("Department")["Salary"].sum()

print(grouped_2)

print(grouped_1)


# for i,v in grouped:
#     print(i)
#     print(v)

# print(df)

print(grouped)
```
