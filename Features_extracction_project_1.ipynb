```python
def extract_episode_count(title):
    eps_cnt=""
    check = False
    for i in title:
    
        if  i == ")":
            check = False
            break
    
        if  i == "(":
            check = True
            continue
            
        if(check == True):
            eps_cnt+=i
             
    return eps_cnt
    
    
```

```python
print(extract_episode_count(df.loc[1].Title))
```

```python
df["Episode_Count"] = df["Title"].apply(extract_episode_count)
```

```python
def extract_Timestamp(title):
    check = False
    timestamp=""
    for ind, val in enumerate(title):
        if val==')':

            check = True           
            continue    
            
        if check == True:
            timestamp = title[ind:ind+19]
            break
            
    return timestamp
```

```python
df["Timestamps"] = df["Title"].apply(extract_Timestamp)

print(df)
```
