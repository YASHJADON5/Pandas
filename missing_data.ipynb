```python
import numpy as np 
import pandas as pd
```

```python
data= {"A": [1,np.nan,3,4], "B": [5,6,7,np.nan], "C": [9,10,11,12]}

df=pd.DataFrame(data, index=["a","b","c","d"])
```

```python
# find missing data in a DataFrame
# removing missing data from a DataFrame
# filling missing data in a DataFrame
```

```python
print(df.isna())
```

```python
print(df.isna().sum())
```

```python
print(df.isna().any())
```

```python
print(df.dropna())
```

```python
df.dropna(thresh=2)
```

```python
df.fillna({"A":0})
```
