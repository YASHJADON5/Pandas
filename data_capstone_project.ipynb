```python
# highest population country

highest_population = df[df["population"]==df["population"].max()]
print(highest_population.country)
# print(df.nlargest(1,"population"))

# highest population capital city 
print(highest_population.capital_city)

# country with the least populaiton 

least_population = df[df["population"]==df["population"].min()]
print(least_population.country)

# least population capital city
print(least_population.capital_city)


# count of unique regions

unique_regions=df["region"].nunique()

# total number of regions

print(unique_regions)

# total countries in eastern europe region

print(df["region"].value_counts()["Eastern Europe"])

print(df[df["region"]=="Eastern Europe"]["country"])


# print(unique_regions.["region"].size())

# unique_regions.to_csv("output4.csv")

# political leader of second highly populated country 


second_highest_populated_country = df.nlargest(2,"population").iloc[1]

print(second_highest_populated_country.political_leader)

```
