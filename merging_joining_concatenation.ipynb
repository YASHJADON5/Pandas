```python
data_1 = {
    "student_id": ["S1", "S2", "S3", "S4"],
    "name": ["Alice", "Bob", "Charlie", "David"],
    "grade": ["A", "B", "C", "D"]
}
data_2 = {
    "student_id": ["S1", "S2", "S3", "S5"],
    "roll-no.": [1, 2, 3, 4]
}

df1 = pd.DataFrame(data_1)
df2 = pd.DataFrame(data_2)
```

```python
print(df1)

print("\n")


print(df2)
```

```python
merged_df = pd.merge(df1,df2, on="student_id", how="inner")
print(merged_df)
```

```python
data_3 = {'A': [1, 2, 3], 'B': [4, 5, 6] , 'C':[7,8,9]}
data_4 = {'A':[70,80,90],'E': [10, 11, 12], 'F': [13, 14, 15] , 'G':[16,17,18]}

df3 = pd.DataFrame(data_3)
df4 = pd.DataFrame(data_4)


concaticated_df  = pd.concat([df3, df4],ignore_index=True)

print(concaticated_df)

print("\n")

print(pd.concat([df3,df4], axis=1 ))
```

```python
# ignore index means reset index, start index from zero for the newly created dataframe
```

```python
# Yes, basically correct. ✅

# Merge → common key is usually a column
# Join → common key is usually the index

# Both are used to combine related DataFrames.
```

```python
data_5 = {'Name':["Bob","Jake","Lisa","Sue"]}
data_6 = {'Grade':["A","B","C","D"],}


df4 = pd.DataFrame(data_5,index=[1,3,4,5])
df5 = pd.DataFrame(data_6)

print(df5)
print()

joined_df = df4.join(df5)

joined_outer_df = df4.join(df5, how="outer")
print(joined_outer_df)
print()
print(joined_df)
```
